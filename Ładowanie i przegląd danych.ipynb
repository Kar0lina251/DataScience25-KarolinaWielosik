```python
import pandas as pd
pd.__version__
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/softhints/Pandas-Exercises-Projects/main/data/food_recipes.csv')

# 2. Wyświetl podstawowe informacje o danych
print("Informacje o danych:")
print(df.info())

print("\nStatystyki opisowe:")
print(df.describe(include='all'))

# 3. Sprawdź liczbę brakujących wartości
print("\nLiczba brakujących wartości w każdej kolumnie:")
print(df.isnull().sum())

# 4. Usuń duplikaty
df_clean = df.drop_duplicates()

print("\nLiczba wierszy przed usunięciem duplikatów:", len(df))
print("Liczba wierszy po usunięciu duplikatów:", len(df_clean))

# 5. Wyświetl 5 pierwszych wierszy oczyszczonego zbioru
print("\nPierwsze 5 wierszy danych:")
print(df_clean.head())
```

```python
#sprawdzam duplikaty
print("Liczba duplikatów:", df.duplicated().sum())
print("Brakujące wartości w kolumnach:")
print(df.isnull().sum())
```

```python
#usuwamy duplikaty
df_clean = df.drop_duplicates()      # usuwa duplikaty
df_clean = df_clean.dropna()         # usuwa wiersze z brakami
print("Nowy rozmiar danych:", df_clean.shape)
df_clean.to_csv("food_recipes_clean.csv", index=False)
df_clean.head()
```
